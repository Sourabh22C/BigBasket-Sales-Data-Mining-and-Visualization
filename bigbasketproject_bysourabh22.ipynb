#Step 8: Find out the outliers from the dataset according to the columns

```python
numeric=df.select_dtypes(include=["int","float"])
numeric
```

```python
def select_outliers(data):
  q1=data.quantile(.25)
  q3=data.quantile(.75)
  iqr=q3-q1
  upper=q3+1.5*iqr
  lower=q1-1.5*iqr
  outliers=data[(data<lower)|(data>upper)]
  return outliers.count()
```

```python
numeric_colm=[	"sale_price"	,"market_price"	,"rating"	,"Discount_percentage"]
outliers_count={col:select_outliers(df[col])for col in numeric_colm}
outliers_count
```

#CLEANING DATA AND REPLACE THE OUTLIERS WITH HELP OF MEAN

```python
def replace_outliers(df,colm):
  q1=df[colm].quantile(.25)
  q3=df[colm].quantile(.75)
  iqr=q3-q1
  upper=q3+1.5*iqr
  lower=q1-1.5*iqr
  df.loc[(df[colm]<lower)|(df[colm]>upper),colm]=df[colm].mean()
  return df
```

#9.1 Number of Products per Category

categories.

Gourmet & World Food ranks second with 4,690 products, followed by Kitchen, Garden & Pets with 3,580 products.

between 2,675 and 2,814.

respectively.

Fruits & Vegetables and Eggs, Meat & Fish have the lowest product counts, at 557 and 350, respectively.
